```python
sales_median = product_analysis['Sales'].median()
margin_median = product_analysis['Gross Margin %'].median()

print("Sales Median:", sales_median)
print("Margin Median:", margin_median)
```

```python
def classify_product(row):
    if row['Sales'] >= sales_median and row['Gross Margin %'] >= margin_median:
        return 'Star Product'
    elif row['Sales'] < sales_median and row['Gross Margin %'] >= margin_median:
        return 'Growth Opportunity'
    elif row['Sales'] >= sales_median and row['Gross Margin %'] < margin_median:
        return 'Margin Risk'
    else:
        return 'Low Priority'

product_analysis['Profitability Category'] = product_analysis.apply(
    classify_product,
    axis=1
)
```

```python
profit_pareto = product_analysis.sort_values(
    'Gross Profit',
    ascending=False
).copy()
```

```python
profit_pareto['Cumulative Profit %'] = (
    profit_pareto['Cumulative Profit'] /
    profit_pareto['Gross Profit'].sum()
) * 100
```

```python
profit_pareto[
    ['Product Name',
     'Gross Profit',
     'Profit Contribution %',
     'Cumulative Profit %']
]
```

```python
products_for_80_profit = (
    profit_pareto['Cumulative Profit %'].searchsorted(80) + 1
)

total_products = len(profit_pareto)

percentage_products = (
    products_for_80_profit / total_products
) * 100

print("Products needed for 80% of profit:", products_for_80_profit)
print("Total products:", total_products)
print("Percentage of products:", percentage_products)
```

```python
margin_risk = product_analysis[
    product_analysis['Gross Margin %'] < product_analysis['Gross Margin %'].median()
].sort_values('Gross Margin %')

margin_risk[
    ['Product Name',
     'Sales',
     'Cost',
     'Gross Profit',
     'Gross Margin %',
     'Cost % of Sales']
]
```

```python
sales_median = product_analysis['Sales'].median()
margin_median = product_analysis['Gross Margin %'].median()

def margin_risk_flag(row):
    if row['Gross Margin %'] < margin_median and row['Sales'] >= sales_median:
        return 'High Priority'
    elif row['Gross Margin %'] < margin_median and row['Sales'] < sales_median:
        return 'Monitor'
    else:
        return 'Healthy'

product_analysis['Margin Risk Flag'] = product_analysis.apply(
    margin_risk_flag,
    axis=1
)
```

```python
product_analysis[
    ['Product Name',
     'Sales',
     'Gross Profit',
     'Gross Margin %',
     'Cost % of Sales',
     'Margin Risk Flag']
].sort_values('Gross Margin %')
```
